### 1.4 Clustering Algorithms to Be Implemented

This project will implement and compare **3 different unsupervised clustering algorithms:**

#### **1. K-Means Clustering**
- **Type:** Partition-based, centroid-based
- **How it works:** Divides data into K clusters by minimizing within-cluster variance
- **Advantages:** Fast, scalable, simple to understand
- **Best for:** Well-separated spherical clusters


#### **2. Hierarchical Clustering (Agglomerative)**
- **Type:** Hierarchical, bottom-up approach
- **How it works:** Builds dendrograms by merging closest clusters
- **Advantages:** Interpretable, no need to specify K beforehand
- **Best for:** Understanding cluster relationships

#### **3. DBSCAN (Density-Based Spatial Clustering)**
- **Type:** Density-based
- **How it works:** Groups points based on density and proximity
- **Advantages:** Can find arbitrary-shaped clusters, identifies outliers
- **Best for:** Non-spherical clusters, noisy data

```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import gaussian_kde

def plot_distribution(df, col, bins=30):
    
    data = df[col].dropna()

    hist = go.Histogram(
        x=data,
        nbinsx=bins,
        histnorm='probability density',
        marker=dict(
            color='#3498DB',
            line=dict(color='black', width=1.5)
        ),
        name='Histogram'
    )

    fig = go.Figure([hist])

    if data.dtype != 'object' and data.nunique() > 2:
        kde = gaussian_kde(data)

        x_range = np.linspace(data.min(), data.max(), 200)
        y_kde = kde(x_range)

        fig.add_trace(
            go.Scatter(
                x=x_range,
                y=y_kde,
                mode='lines',
                line=dict(color='red', width=2),
                name='KDE Curve'
            )
        )

    fig.update_layout(
        title=f'Distribution of {col}',
        xaxis_title=col,
        yaxis_title='Density',
        width=650,
        height=400,
        template='plotly_white'
    )

    fig.show()
```

## 5. Model Selection

### 5.1 K-Means Clustering

K-Means is a partition-based clustering algorithm that divides
data into K clusters by minimizing within-cluster variance. It is
the most widely used clustering algorithm in industry.

**How it works:**
- Randomly initialize K centroids
- Assign each point to the nearest centroid
- Recalculate centroids based on assigned points
- Repeat until convergence

### 5.2 Finding the Optimal Number of Clusters

Before applying K-Means, we need to determine the ideal value of K.
We use two methods for this:

### 5.3 Elbow Method

The Elbow Method plots inertia (within-cluster sum of squares)
against different values of K. The point where the curve bends
like an "elbow" indicates the optimal number of clusters.

**How it works:**
- Run K-Means for K = 1 to 10
- Calculate inertia for each K
- Plot inertia vs K
- Identify the elbow point where inertia stops decreasing significantly

```python
from sklearn.cluster import KMeans

inertia = []

K = range(1,11)
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42,n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

inertia
```

```python
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=list(K),
    y=inertia,
    mode='lines+markers',
    marker=dict(
        color='#E74C3C',
        size=8,
        line=dict(color='black', width=1)
    ),
    line=dict(color='#3498DB', width=2),
    hovertemplate='<b>K:</b> %{x}<br><b>Inertia:</b> %{y:.2f}<extra></extra>'
))

fig.update_layout(
    title='Elbow Method - Optimal K Selection',
    xaxis_title='Number of Clusters (K)',
    yaxis_title='Inertia',
    width=645,
    height=400,
    template='plotly_white'
)
fig.show()
```

### 5.4 Silhouette Analysis

Silhouette Analysis measures how similar a point is to its own
cluster compared to other clusters. Score ranges from -1 to 1 —
higher is better.

**How it works:**
- Run K-Means for K = 2 to 10
- Calculate silhouette score for each K
- Higher score = better defined clusters
- Select K with the highest silhouette score

```python
from sklearn.metrics import silhouette_score

sil_scores = []

K = range(2,11)
for k in K:
    kmeans = KMeans(n_clusters=k,random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    sil_scores.append(score)

sil_scores
```

```python
import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=list(K),
        y=sil_scores,
        mode='lines+markers',
        line=dict(color='deepskyblue', width=3),
        marker=dict(size=10, color='red'),
        name='Silhouette Score'
    )
)

fig.update_layout(
    title='Silhouette Analysis',
    xaxis_title='Number of Clusters (k)',
    yaxis_title='Silhouette Score',
    template='plotly_white',
    hovermode='x unified',
    width=645,
    height=400,
)

fig.show()
```

### 5.5 Decision — Final K Selection

Based on both the Elbow Method and Silhouette Analysis, we select
the optimal number of clusters.

**Elbow Method:** Curve bends at K = 5, indicating diminishing
returns beyond this point.

**Silhouette Analysis:** K = 5 yields the highest silhouette score,
confirming well-defined and separated clusters.

**Final Decision: K = 5**

K = 5 will be used for all subsequent clustering algorithms.

### 6.4 Customer Segmentation

Each cluster is ranked based on business value using
AnnualIncome and SpendingScore. Higher income and higher
spending = more valuable customer.

**BusinessScore Formula:**
- 50% AnnualIncome + 50% SpendingScore
- Age excluded (age does not determine customer value)

```python
cluster_stats = df.groupby("KMeans_Cluster")[["AnnualIncome", "SpendingScore", "Age"]].mean().reset_index()

norm = cluster_stats.copy()
for col in ["AnnualIncome", "SpendingScore", "Age"]:
    norm[col] = (norm[col] - norm[col].min()) / (norm[col].max() - norm[col].min())

norm["BusinessScore"] = (
    0.50 * norm["AnnualIncome"] +
    0.50 * norm["SpendingScore"]
)

norm = norm.sort_values("BusinessScore", ascending=False).reset_index(drop=True)

labels = [
    ("⭐⭐⭐⭐⭐", "Best"),
    ("⭐⭐⭐⭐", "Potential"),
    ("⭐⭐⭐", "Sustainable"),
    ("⭐⭐", "High Risk"),
    ("⭐", "Low Value")
]

norm["Business Value"] = [l[0] for l in labels]
norm["Behavior"] = [l[1] for l in labels]

final_table = norm[["KMeans_Cluster", "Business Value", "Behavior"]] \
    .rename(columns={"KMeans_Cluster": "Cluster"})

final_table
```

## 7. Hierarchical Clustering (Agglomerative)

### 7.1 Introduction

Hierarchical Clustering builds a tree of clusters (dendrogram)
by progressively merging the closest data points. Unlike K-Means,
it does not require specifying K beforehand.

**How it works:**
- Start with each point as its own cluster
- Merge the two closest clusters
- Repeat until all points are in one cluster
- Cut the dendrogram at desired level to get K clusters

```python
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
linked = linkage(X_scaled, method='ward')
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage

linked = linkage(X_scaled, method='ward')

dend = dendrogram(linked, truncate_mode='lastp', p=12, no_plot=True)

color_map = {
    'C0': '#3498DB',
    'C1': '#E74C3C',
    'C2': '#2ECC71',
    'C3': '#F39C12',
    'C4': '#9B59B6',
    'C5': '#1ABC9C',
    'b': '#3498DB',
    'g': '#2ECC71',
    'r': '#E74C3C',
    'k': '#2C3E50'
}

fig = go.Figure()

for icoord, dcoord, color in zip(dend['icoord'], dend['dcoord'], dend['color_list']):
    plotly_color = color_map.get(color, '#3498DB')
    fig.add_trace(go.Scatter(
        x=icoord,
        y=dcoord,
        mode='lines',
        line=dict(color=plotly_color, width=2),
        hovertemplate='<b>Distance:</b> %{y:.2f}<extra></extra>',
        showlegend=False
    ))

tick_text = [f"({dend['ivl'][i]})" for i in range(len(dend['ivl']))]

fig.update_layout(
    title='Hierarchical Clustering Dendrogram',
    xaxis=dict(
        tickvals=[5 + 10*i for i in range(len(tick_text))],
        ticktext=tick_text,
        title='Cluster Size'
    ),
    yaxis_title='Distance',
    template='plotly_white',
    width=700,
    height=400
)

fig.show()
```

## 8. DBSCAN Clustering


DBSCAN (Density-Based Spatial Clustering of Applications with Noise)
groups points based on density. Unlike K-Means, it does not require
specifying K and can detect outliers as noise points.

**Key Parameters:**
- **eps** — Maximum distance between two points to be considered neighbors
- **min_samples** — Minimum points required to form a dense region

**Advantages:**
- Finds arbitrary-shaped clusters
- Automatically detects outliers (noise points)
- No need to specify number of clusters

```python
for eps in [0.1, 0.15, 0.2, 0.25, 0.3]:
    db = DBSCAN(eps=eps, min_samples=5)
    labels = db.fit_predict(X_scaled)
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    n_noise = list(labels).count(-1)
    print(f"eps={eps} | Clusters: {n_clusters} | Noise: {n_noise}")
```

```python
dbscan = DBSCAN(eps=0.2, min_samples=5)
labels_dbscan = dbscan.fit_predict(X_scaled)
df["DBSCAN_Cluster"] = labels_dbscan

print("Unique Labels:", np.unique(labels_dbscan))
print("Clusters:", len(set(labels_dbscan)) - (1 if -1 in labels_dbscan else 0))
print("Noise Points:", list(labels_dbscan).count(-1))
```

```python
fig = go.Figure()

for label in np.unique(labels_dbscan):
    if label == -1:
        fig.add_trace(go.Scatter(
            x=X[labels_dbscan == label, 0],
            y=X[labels_dbscan == label, 1],
            mode='markers',
            name='Noise',
            marker=dict(color='black', size=6, symbol='x'),
            hovertemplate='<b>Noise</b><br><b>Annual Income:</b> %{x}k$<br><b>Spending Score:</b> %{y}<extra></extra>'
        ))
    else:
        fig.add_trace(go.Scatter(
            x=X[labels_dbscan == label, 0],
            y=X[labels_dbscan == label, 1],
            mode='markers',
            name=f'Cluster {label}',
            marker=dict(size=8),
            hovertemplate=f'<b>Cluster {label}</b><br><b>Annual Income:</b> %{{x}}k$<br><b>Spending Score:</b> %{{y}}<extra></extra>'
        ))

fig.update_layout(
    title='DBSCAN Clustering (eps=0.2)',
    xaxis_title='Annual Income (k$)',
    yaxis_title='Spending Score (1-100)',
    template='plotly_white',
    hovermode='closest',
    width=650,
    height=400
)

fig.show()
```

## 9. Model Comparison
### 9.1 Silhouette Score Comparison

We compare all clustering algorithms based on Silhouette Score
and number of clusters to determine the best performing model
for mall customer segmentation.



```python
from time import time

results = []

# K-Means
t = time()
km = KMeans(n_clusters=5, random_state=42, n_init=10, init='random')
l = km.fit_predict(X_scaled)
results.append({
    "Algorithm": "K-Means",
    "N Clusters": 5,
    "Silhouette": round(silhouette_score(X_scaled, l), 4),
    "Runtime (s)": round(time() - t, 3)
})


# Hierarchical
t = time()
h = AgglomerativeClustering(n_clusters=5, linkage='ward')
l = h.fit_predict(X_scaled)
results.append({
    "Algorithm": "Hierarchical",
    "N Clusters": 5,
    "Silhouette": round(silhouette_score(X_scaled, l), 4),
    "Runtime (s)": round(time() - t, 3)
})

# DBSCAN
t = time()
db = DBSCAN(eps=0.2, min_samples=5)
l = db.fit_predict(X_scaled)
mask = l != -1
n_clusters = len(set(l)) - (1 if -1 in l else 0)
sil = round(silhouette_score(X_scaled[mask], l[mask]), 4) if n_clusters > 1 else "N/A"
results.append({
    "Algorithm": "DBSCAN",
    "N Clusters": n_clusters,
    "Silhouette": sil,
    "Runtime (s)": round(time() - t, 3)
})



df_results = pd.DataFrame(results)
```

### 9.2 Additional Evaluation Metrics and Scalability

The earlier comparison used silhouette score and runtime. To strengthen the evaluation, we add Calinski-Harabasz and Davies-Bouldin scores to assess cluster separation and compactness. We also benchmark runtime across different sample sizes to discuss scalability.

```python
from time import perf_counter


def compute_metrics(labels, X, algorithm_name):
    mask = labels != -1 if algorithm_name == "DBSCAN" else np.ones(len(labels), dtype=bool)
    X_use = X[mask]
    labels_use = labels[mask]

    if len(np.unique(labels_use)) < 2:
        return {
            "Algorithm": algorithm_name,
            "N Clusters": len(np.unique(labels_use)),
            "Silhouette": np.nan,
            "Calinski-Harabasz": np.nan,
            "Davies-Bouldin": np.nan,
        }

    return {
        "Algorithm": algorithm_name,
        "N Clusters": len(np.unique(labels_use)),
        "Silhouette": round(silhouette_score(X_use, labels_use), 4),
        "Calinski-Harabasz": round(calinski_harabasz_score(X_use, labels_use), 2),
        "Davies-Bouldin": round(davies_bouldin_score(X_use, labels_use), 2),
    }


evaluation_metrics = pd.DataFrame([
    compute_metrics(labels_kmeans, X_scaled, "K-Means"),
    compute_metrics(labels_hier, X_scaled, "Hierarchical"),
    compute_metrics(labels_dbscan, X_scaled, "DBSCAN"),
])

evaluation_metrics
```

```python
for metric in ["Calinski-Harabasz", "Davies-Bouldin"]:
    plot_df = evaluation_metrics[["Algorithm", metric]].copy().dropna()

    if plot_df.empty:
        continue

    fig = px.bar(
        plot_df,
        x="Algorithm",
        y=metric,
        color="Algorithm",
        title=f"{metric} Comparison",
        labels={metric: metric},
        template="plotly_white",
        text=plot_df[metric].round(2)
    )
    fig.update_traces(textposition="outside")
    fig.update_layout(
        showlegend=False,
        width=640,
        height=400,
        margin=dict(t=50, l=20, r=20, b=20),
        yaxis=dict(range=[0, max(plot_df[metric].max() * 1.15, 1.0)])
    )
    fig.show()
```

```python
sample_sizes = [200, 500, 1000]
scalability_results = []

for n in sample_sizes:
    X_sub = X_scaled[:n]
    for algorithm_name, model in [
        ("K-Means", KMeans(n_clusters=5, random_state=42, n_init=10)),
        ("Hierarchical", AgglomerativeClustering(n_clusters=5)),
        ("DBSCAN", DBSCAN(eps=0.2, min_samples=5)),
    ]:
        start = perf_counter()
        model.fit_predict(X_sub)
        runtime = perf_counter() - start
        scalability_results.append({
            "Algorithm": algorithm_name,
            "Sample Size": n,
            "Runtime (s)": round(runtime, 4),
        })

scalability_df = pd.DataFrame(scalability_results)
scalability_df
```
